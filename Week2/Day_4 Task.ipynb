# Merge two small Datasets

```python
import pandas as pd
```

```python
print("merging datasets using pandas")
```

```python
students = pd.read_csv("students.csv")
city = pd.read_csv("city.csv")
```

```python
print("Students Data")
print(students)
```

```python
print("Students Address/city")
print(city)
```

>> Merge DataSets

```python
merged = pd.merge(students, city, on="ID")
print("\nMerged Dataset")
print(merged)
```

>> Students statistics

```python
print("\n Summary")
print("Total students : ",len(merged))
print("Average Marks : ",merged["Marks"].mean())
print("Higest Marks : ",merged["Marks"].max())
print("Lowest Marks : ",merged["Marks"].min())
```

>> Department wise

```python
dept_avg = merged.groupby("Department")["Marks"].mean()
print(dept_avg)
```

>> Grade columns

```python
def grade(mark):

    if mark >= 90:
        return "A+"

    elif mark >= 80:
        return "A"

    elif mark >= 70:
        return "B"

    elif mark >= 60:
        return "C"

    elif mark >= 50:
        return "D"

    else:
        return "F"


merged["Grade"] = merged["Marks"].apply(grade)

print("\nDataset with Grades")

print(merged)
```

>> Save OutPUT

```python
merged.to_csv(
    "merged_students.csv",
    index=False
)

print("\nMerged dataset saved successfully!")

print("\nTask Completed Successfully!")
```
