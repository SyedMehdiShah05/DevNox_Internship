> EDA Mini-Dashboard Script — load a public dataset (or fetched API data), auto-generate 4 charts, and write 3–4 sentences of insight based on what the charts show.


```python
import os

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

```

```python

# 1. Create charts directory

os.makedirs("Task7_charts", exist_ok=True)

 # 2. Load Dataset
df = sns.load_dataset("iris")
```

```python

# 3. Basic EDA

print("\nEDA MINI DASHBOARD")
print("=" * 70)

print("\nFirst 5 Rows:")
print(df.head())
print("=" * 70 )

print("\nDataset Shape:")
print(df.shape)
print("=" * 40 )

print("\nDataset Information:")
print(df.info())
print(" =" * 50 )

print("\nMissing Values:")
print(df.isnull().sum())
print("=" * 50 )

print("\nSummary Statistics:")
print(df.describe())
print("=" * 80 )
```

```python
# 4. Handle Missing Values

numeric_columns = df.select_dtypes(include="number").columns

df[numeric_columns] = df[numeric_columns].fillna(
    df[numeric_columns].mean()
)
```

```python
# CHART 1: Histogram

plt.figure(figsize=(8, 5))

sns.histplot(
    data=df,
    x="petal_length",
    bins=20,
    kde=True
)

plt.title("Distribution of Petal Length")
plt.xlabel("Petal Length")
plt.ylabel("Frequency")

plt.tight_layout()

plt.savefig(
    "Task7_charts/histogram.png",
    dpi=300
)

plt.close()

```

```python
# CHART 2: Bar Chart

species_count = df["species"].value_counts()

plt.figure(figsize=(8, 5))

species_count.plot(kind="bar")

plt.title("Number of Flowers by Species")
plt.xlabel("Species")
plt.ylabel("Count")

plt.xticks(rotation=0)

plt.tight_layout()

plt.savefig(
    "Task7_charts/bar_chart.png",
    dpi=300
)

plt.close()

```

```python
# CHART 3: Scatter Plot

plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="petal_length",
    y="petal_width",
    hue="species"
)

plt.title("Petal Length vs Petal Width")
plt.xlabel("Petal Length")
plt.ylabel("Petal Width")

plt.tight_layout()

plt.savefig(
    "Task7_charts/scatter_plot.png",
    dpi=300
)

plt.close()

```

```python
# CHART 4: Correlation Heatmap

correlation = df.select_dtypes(
    include="number"
).corr()

plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Feature Correlation Heatmap")

plt.tight_layout()

plt.savefig(
    "Task7_charts/correlation_heatmap.png",
    dpi=300
)

plt.close()
```

```python
# 5. Generate Insights

import numpy as np
strongest_correlation = (
    correlation
    .where(np.eye(correlation.shape[0], dtype=bool))
    .abs()
    .stack()
    .idxmax()
)

print("\n" + "=" * 50)
print("EDA INSIGHTS")
print("=" * 50)

insights = [
    "The dataset contains three different flower species with an equal number of observations.",
    "Petal length shows different distributions across the flower species.",
    "Petal length and petal width have a strong positive relationship.",
    "The scatter plot shows that the flower species can be visually separated using petal measurements."
]

for i, insight in enumerate(insights, 1):
    print(f"{i}. {insight}")

```

```python
# 6. Save Insights

with open("Task7_insights.txt", "w", encoding="utf-8") as file:

    for i, insight in enumerate(insights, 1):
        file.write(f"{i}. {insight}\n")


print("\nCharts saved in: charts/")
print("Insights saved in: Task7_insights.txt")
print("\nEDA Dashboard completed successfully!")
```
