# Task 4 JSON API
>Parse a JSON API response and turn part of it into a chart.

```python
import json

with open("users.json") as file:
    data = json.load(file)
```

```python
companies = []

for user in data:
    companies.append(user["company"]["name"])

print(companies)
```

```python
import matplotlib.pyplot as plt

company_lengths = [len(name) for name in companies]

plt.bar(companies, company_lengths)

plt.xticks(rotation=90)

plt.ylabel("Characters")

plt.title("Company Name Length")



plt.show()
```
