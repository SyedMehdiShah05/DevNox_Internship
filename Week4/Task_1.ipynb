- > Task 1:  Compute mean, median, standard deviation, and variance by hand, then verify with NumPy.

```python
import numpy as np
```

```python
data = [56,78,23,45,889,12,54,44,1]

sorted_data = np.sort(data) # sorte data
print("Sorted Data:", sorted_data)

length = len(data) # length of data
print("Length:", length)
```

```python

mean = np.mean(data) # mean_formula = sum(data) / length
print("Mean:", mean)

median = np.median(data) # median
print("Median:", median)

mode = np.bincount(data).argmax() # mode
print("Mode:", mode)

variance = np.var(data) # varience_formula = sum((x - mean) ** 2 for x in data) / length
print("Variance:", variance)

std_dev = np.std(data) # std_dev_formula = variance ** 0.5
print("Standard Deviation:", std_dev)

data_range = np.ptp(data) # range_formula = max(data) - min(data)
print("Range:", data_range)

max_value = np.max(data) # max_value
print("Max Value:", max_value)

min_value = np.min(data) # min_value
print("Min Value:", min_value) 
```

```python

# numpyVerification
print("\nNumpy Verification:")

np.mean(data) == mean
print("\nMean Verification:", np.mean(data) == mean)

np.median(data) == median
print("Median Verification:", np.median(data) == median)

np.bincount(data).argmax() == mode
print("Mode Verification:", np.bincount(data).argmax() == mode)

np.var(data) == variance
print("Variance Verification:", np.var(data) == variance)

np.std(data) == std_dev
print("Standard Deviation Verification:", np.std(data) == std_dev)

np.ptp(data) == data_range
print("Range Verification:", np.ptp(data) == data_range)

np.max(data) == max_value
print("Max Value Verification:", np.max(data) == max_value)

np.min(data) == min_value
print("Min Value Verification:", np.min(data) == min_value)
```
