```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def sigmoid(z):
    return 1 / (1 + np.exp(-z))


# Test values
values = np.array([-10, -5, -2, -1, 0, 1, 2, 5, 10])

results = sigmoid(values)

for z, result in zip(values, results):
    print(f"z = {z:>3}  sigmoid = {result:.4f}")

```

```python

# Plot sigmoid
z = np.linspace(-10, 10, 100)

y = sigmoid(z)

plt.plot(z, y)

plt.xlabel("z")
plt.ylabel("sigmoid(z)")
plt.title("Sigmoid Function")

plt.grid()
plt.show()
```
