# Task_2
- > Implement simple linear regression from scratch using manual gradient descent.


```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Dataset

X = np.array([1, 2, 3, 4, 5, 6, 7, 8], dtype=float)
y = np.array([35, 40, 50, 55, 65, 70, 75, 85], dtype=float)


# Initial parameters

w = 0.0
b = 0.0

learning_rate = 0.01
epochs = 2000

n = len(X)
```

```python

# Gradient Descent

for epoch in range(epochs):

    # Prediction
    y_pred = w * X + b

    # Error
    error = y_pred - y

    # Gradients
    dw = (2 / n) * np.sum(X * error)
    db = (2 / n) * np.sum(error)

    # Update parameters
    w = w - learning_rate * dw
    b = b - learning_rate * db
```

```python
# Final model

print("Weight (w):", w)
print("Bias (b):", b)

print("\nEquation:")
print(f"Marks = {w:.2f} × Study Hours + {b:.2f}")
```

```python
# Prediction

study_hours = 7

predicted_marks = w * study_hours + b

print(
    f"\nPredicted marks for {study_hours} study hours:",
    predicted_marks
)
```

```python
# Visualization
 
plt.scatter(X, y, label="Actual Data")

plt.plot(
    X,
    w * X + b,
    label="Regression Line"
)

plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.title("Linear Regression From Scratch")

plt.legend()
plt.show()
```
