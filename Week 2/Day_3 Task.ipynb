# Task 3 Pandas

```python
import pandas as pd
```

```python
df = pd.read_csv("D:\Messy_Employee_dataset.csv")
```

```python
head_df = df.head(10)
print(head_df)
```

```python
# Check for missing values in each column
missing_values = df.isnull().sum()
print(missing_values)
```

### Data Cleaning Steps

1.  **Handle Missing Values**: Impute missing 'Age' and 'Salary' values with their respective medians.
2.  **Convert Data Types**: Convert 'Join_Date' to datetime objects.

```python
# Impute missing 'Age' values with the median
age_median = df['Age'].median()
df['Age'].fillna(age_median, inplace=True)

print(f"Missing 'Age' values imputed with median: {age_median}")
```

```python
# Impute missing 'Salary' values with the median
salary_median = df['Salary'].median()
df['Salary'].fillna(salary_median, inplace=True)

print(f"Missing 'Salary' values imputed with median: {salary_median}")
```

```python
# Convert 'Join_Date' to datetime objects
df['Join_Date'] = pd.to_datetime(df['Join_Date'], errors='coerce')

print(" 'Join_Date' column converted to datetime.")
```

### Verification of Cleaning Steps

```python
# Recheck for missing values after imputation
print("Missing values after imputation:")
print(df.isnull().sum())

# Check data types after conversion
print("\nData types after conversion:")
print(df.info())
```
