# 1. Bar Chart

```python
import matplotlib.pyplot as plt
```

```python
month = ["jan", "feb", "mar", "apr", "may", "jun", "jul", "aug", "sep", "oct", "nov", "dec"]
sales = [1000, 1200, 1500, 1700, 2000, 2500, 3000, 3500, 4000, 4500, 5000, 6000]
```

```python
plt.figure(figsize=(8, 5))
plt.bar(month, sales, color='red')
plt.title("Monthly Sales")

plt.xlabel("Month")
plt.ylabel("Sales")
plt.show()
```

# 2. Line Chart

```python
import matplotlib.pyplot as plt
```

```python
months = ["jan", "feb", "mar", "apr", "may", "jun", "jul", "aug", "sep", "oct", "nov", "dec"]
sales = [1000, 200, 1500, 1700, 200, 2500, 300, 3500, 400, 400, 5000, 600]
```

```python
plt.figure(figsize=(10, 5))
plt.plot(months, sales, marker='o', color='green')
plt.title("Monthly Sales Trend")

plt.xlabel("Month")
plt.ylabel("Sales")

plt.grid(True)
plt.show()
```

# 3. Pie Chart

```python
plt.figure(figsize=(10, 5))

plt.pie(
    sales,
    labels=months,
    autopct='%1.1f%%',
    startangle=90,
)

plt.title("Sales Distribution by Month")
plt.show()
```
