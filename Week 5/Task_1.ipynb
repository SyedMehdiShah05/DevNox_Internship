> - Task 1 : Implement K-Nearest Neighbors from scratch on a small 2D dataset.

```python
import numpy as np
from collections import Counter
```

```python
X = np.array([
    [160, 55],
    [165, 60],
    [170, 65],
    [175, 70],
    [180, 75],
    [185, 80]
])
```

```python
y = np.array([0, 0, 0, 1, 1, 1])


def euclidean_distance(point1, point2):
    return np.sqrt(np.sum((point1 - point2) ** 2))


def knn_predict(X_train, y_train, X_test, k=3):
    predictions = []

    for test_point in X_test:

        distances = []

        for i, train_point in enumerate(X_train):
            distance = euclidean_distance(test_point, train_point)
            distances.append((distance, y_train[i]))

        distances.sort(key=lambda item: item[0])

        nearest_neighbors = distances[:k]

        labels = [label for _, label in nearest_neighbors]

        prediction = Counter(labels).most_common(1)[0][0]

        predictions.append(prediction)

    return np.array(predictions)


# This point is clearly in the class-1 region of the training data.
X_test = np.array([[182, 77]])

prediction = knn_predict(X, y, X_test, k=3)

assert prediction[0] == 1
print("Prediction:", prediction)
```
